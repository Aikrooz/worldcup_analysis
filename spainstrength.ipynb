```python
import pandas as pd 
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
attacking_data = pd.read_csv("worldcup_2026_team_attacking_stats.csv").drop("Rank", axis=1)
defensive_data = pd.read_csv("worldcup_2026_team_defensive_stats.csv").drop("Rank", axis=1)
movement_data = pd.read_csv("worldcup_2026_team_offers_receptions_stats.csv").drop("Rank", axis=1)
passing_data = pd.read_csv("worldcup_2026_team_passing_stats.csv").drop("Rank", axis=1)
physical_data = pd.read_csv("worldcup_2026_team_physical_stats.csv").drop("Rank", axis=1)
matches_data=pd.read_csv('worldcupd_marches_played.csv')
```

```python
merged_data = (
    attacking_data
    .merge(defensive_data, on="Team")
    .merge(movement_data, on="Team")
    .merge(passing_data, on="Team")
    .merge(physical_data, on="Team")
    .merge(matches_data,on='Team')
)
```

```python
merged_data['turnovers per match']=round(merged_data['Forced turnovers']/merged_data['Matches Played'])
merged_data['defensive pressured applied per match']=round(merged_data['Defensive pressures applied']/merged_data['Matches Played'])
```

```python
plt.figure(figsize=(12,10))
ax=sns.scatterplot(data=merged_data,x='turnovers per match',y='defensive pressured applied per match')
for i in range(len(merged_data)):
    ax.text(
        merged_data['turnovers per match'][i],
        merged_data['defensive pressured applied per match'][i],
        merged_data['Team'][i],
        fontsize=9,
        ha='left',
        va='bottom'
    )
plt.title("Defensive pressured applied To turnovers won")
plt.xlabel('Turn overs ')
plt.ylabel('Defensive pressure')
```

```python
merged_data['Ball recovery time per match']=round(merged_data['Ball recovery time']/merged_data['Matches Played'])
```

```python
plt.figure(figsize=(12,10))
ax=sns.scatterplot(data=merged_data,x='Ball recovery time per match',y='defensive pressured applied per match')
for i in range(len(merged_data)):
    ax.text(
        merged_data['Ball recovery time per match'][i],
        merged_data['defensive pressured applied per match'][i],
        merged_data['Team'][i],
        fontsize=9,
        ha='left',
        va='bottom'
    )
plt.title("Defensive pressured applied To Ball recovery time")
plt.xlabel('Ball recovery Time ')
plt.ylabel('Defensive pressure')
```
